Day 6 : Loops in python

for Loop

```python
for i in range (5):
  print(i)
```

range function

```python
print(list(range(5)))
print(list(range(2,10)))
```

loop thtough a string

```python
name = "vaibhav"
for ch in name:
  print(ch)
```

Loop through a list

```python
furits = ["apple","banana","orange","gauava"]
for furit in furits:
  print(furit)
```

while loop

```python
i = 1
while i<=5:
  print (i)
  i += 1
```

break

```python
for i in range(1,10):
  if i ==6:
    break
  print(i)
```

continue

```python
for i in range(1,6):
  if i ==3:
    continue
  print(i)
```

pattern printing

```python
for i in range(5):
  print("*"*(i+1))
```

sum of first n number

```python
n = int(input("Enter the no "))
total = 0
for i in range(1,n+1):
 total += i
print (total)
```

Prime number check

```python
num = int(input("enter the no"))
is_prime = True
if num <2:
  is_prime = False
else:
  for i in range(2,int(num**0.5)+1):
   if num  % i ==0:
     is_prime = False
     break
  if is_prime:
    print("prime")
  else:
   print("not a prime")
```
