find the sum of n first nymber

```python
n = int(input("enter the no"))
sum = 0

for i in range(1,n+1):
  sum += i
print(sum)
```

print  the multiplaction of a table

```python
num = int(input("enter the no"))
for i in range(1,11):
  print(f"{num}x{i} = {num*i}")
```

Find the factorial of num

```python
n = int(input('enter the no'))
fact = 1
for i in range(1,n+1):
  fact *= i
print(fact)
```

print the Fibonacci series

```python
n = int(input("enter the no"))
a = 0
b = 1
for i in range(1,n+1):
  print(a,end ="")
  a,b = b,a+b
```

reverse a number

```python
n = int(input("enter the no"))
revese = 0
while n >0:
  digit = n%10
  revese  = revese*10+digit
  n//=10

print(revese)
```
