1. What is Broadcasting

Broadcasting automatically expands the smaller array to match the larger array's shape during operations.

```python
import numpy as np
arr = np.array([1,2,3,4,5,6,7])
print(arr+10)
```

2. Broadcasting with scaler

```python
import numpy as np
arr = np.array([1,2,3])
print(arr+10)
print(arr*2)
```

3. Broadcasting between arrays

```python
import numpy as np
arr1 = np.array([10,20,30])
arr2 = np.array([40,50,60])
print(arr1+arr2)
```

4. Broadcasting in 2D array

```python
import numpy as np
arr = np.array([
    [1,2,3],
    [4,5,6]
])
print(arr+10)
```

5. Broadcasting Row-wise

```python
import numpy as np
matrix = np.array([
    [1,2,3],
    [4,5,6]
])
row = np.array([10,20,30])
print(matrix + row)
```

6. Broadcasting Coloumn- wise  

```python
import numpy as np

matrix = np.array([
    [1, 2, 3],
        [4, 5, 6]
        ])
column = np.array([[10],
                   [20]])

print(matrix + column)
```

7. Broadcasting Rules

Two dimensions are compatible if:
They are equal, or
One of them is 1

Example(works)

(2,3)
(1,3)

Example(works)

(2,3)
(2,1)

Example(error)

(2,3)
(4,3)

8. Broadcasting with multiplication

```python
import numpy as np

arr = np.array([[1, 2],
                [3, 4]])
print(arr * 5)
```
