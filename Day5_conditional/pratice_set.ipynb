Build a simple using calculator using if elif else

```python
num1 = int(input("enter the no "))
op = input("enter the operator +,-,*,/")
num2 = int(input("enter the no"))

if op == "+":
  print("+",num1+num2)
elif op == "-":
  print("-",num1-num2)
elif op == "*":
  print("*",num1*num2)
elif op == "/":
  print("/",num1/num2)
else:
  print("invalid user")

```
