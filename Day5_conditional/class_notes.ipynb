```python
age = 20
if age >= 18 :
  print("egible to vote")
```

if...else

```python
age = 20
if age >= 18:
  print("egible to vote")
else:
  print("not egible to vote")
```

if..elif..else

```python
mark = 70
if mark >= 80:
  print("A+")
elif mark >= 75:
  print("A")
else:
  print("B")
```

4. Nested if

An if statement inside anthoer if statement  

```python
age = 20
ciziten = True
if age >= 18:
  if ciziten:
    print("egible to vote")
  else:
    print("Not a citizen")
else:
    print("undegree")
```

Tarnary  Operator

```python
age = 20
print("Adult") if age >= 18 else print("Minor")
```

Logical operator

```python
age = 22
has_id = True

if age >= 18 and has_id:
    print("Entry Allowed")
```
