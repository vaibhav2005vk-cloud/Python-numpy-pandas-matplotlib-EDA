# Assignment

Write a single Python program that:

° Take two number from the user
° print all arthmatic Opeartion
° print all compraison expression
° calcualte(a+b)*2
° Uses at least one logical operator

```python
num1 = int(input("Enter the first no:")) # first num
num2 = int(input("Enter the second no:")) # second number
#Arthmatic opeartions
print(" 1.Arthmatic operation")
print(num1 + num2) # Addition
print(num1 -num2) # Subtraction
print(num1 * num2) # Multiplication
print(num1 / num2) # Divison
print(num1 // num2) # Floor Divison
print(num1 % num2) # Modulus
# Compraison operator
print(" 2. Comprasion Operator")
print(num1 == num2) # equal to
print(num1 > num2) # greater than
print(num1 < num2) # less than
print(num1 != num2) # not equal to
print(num1 >= num2) # greater than equal
print(num1 <= num2) # less than equal to
# Caluclate(a+b)*2
print(" 3. Calucate(a+b)*2")
print((num1+num2)*2)
```
