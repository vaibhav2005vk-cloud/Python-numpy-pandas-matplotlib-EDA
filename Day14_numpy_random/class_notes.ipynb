6. np.random.seed()

```python
import numpy as np
np.random.seed(42)
print(np.random.randint(1,100,5))
```

7. np.random.shuffle

Shuffle an array in place

```python
import numpy as np
arr = np.array([1, 2, 3, 4, 5])

np.random.shuffle(arr)

print(arr)
```

8. np.random.permutaion

Returns a shuffled copy without changing the original array.

```python

import numpy as np
arr = np.array([1, 2, 3, 4, 5])

new_arr = np.random.permutation(arr)

print(new_arr)
print(arr)
```
