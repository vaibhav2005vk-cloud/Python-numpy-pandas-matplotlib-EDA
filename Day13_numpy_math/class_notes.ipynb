Day 13 Numpy Mathmatical

1. Create an array

```python
import numpy as np
arr = np.array([10,20,30])
```

2. Sum

```python
print(np.sum(arr))
```

Output

150

3. Product

```python
print(np.prod(arr))
```

Output

12000000

4. Mean (Average)

```python
print(np.mean(arr))
```

Output

30.0

5. Median

```python
print(np.median(arr))
```

Output

30.0

6. Standard Deviation

```python
print(np.std(arr))
```

7. Variance

```python
print(np.var(arr))
```

8. Minimum and Maximum

```python
print(np.max(arr))
print(np.min(arr))
```

9. Index of maximum and minimum

```python
print(np.argmin(arr))
print(np.argmax(arr))
```

10. Cumulative Sum

```python
print(np.cumsum(arr))
```

Output

[10,30,60,100,150]

11. Cumulative Product

```python
print(np.cumprod(arr))
```

12. Square root

```python
print(np.sqrt(arr))
```

13. Power

```python
print(np.power(arr,2))
```

14. axis parameter

sum of each colum(axis = 0)

```python
print(np.sum(arr,axis= 0))
```

Output

[5,7,9]

sum of each row(axis=1)

```python
print(np.sum(arr,axis=1))
```

Output

[6,15]

15. Maximum and Minimum

```python
print(np.max(arr,axis=0))
print(np.min(arr,axis=1))
```
