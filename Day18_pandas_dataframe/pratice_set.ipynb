Pratice set

1. Create a DataFrame with Name, Age, and Marks.
2. Print the shape.
3. Print the column names.
4. Print the data types.
5. Print the first 2 rows.
6. Print the last row.
7. Print only the Marks column.
8. Print Name and Marks together.
9. Add a Grade column.
10. Delete the Grade column.

```python
import pandas as pd
data ={
   "Name" : ["vaibhav","Abhay","Ayush"],
   "Age" : [24,21,14],
   "Marks" : [80,90,75]
}
df = pd.DataFrame(data)
print(df)
print(df.shape)
print(df.columns)
print(df.dtypes)
print(df.head(2))
print(df.tail())
print(df["Marks"])
print(df[["Name","Marks"]])
df["Grade"] = ["A","B","C"]
print(df)
df.drop("Grade",axis = 1,inplace = True)
print(df)
```
