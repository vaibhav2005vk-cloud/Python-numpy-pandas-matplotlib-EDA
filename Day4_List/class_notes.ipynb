3. Acessing Elements

```python
furits = ["apple","banana","cherry","gauava"]
print(furits[0])
print(furits[-1])
print(furits[1])
```

4. slicing

```python
numbers = [10,20,30,40,50,60,70]
print(numbers[1:4])
print(numbers[:4])
print(numbers[4:])
print(numbers[::-1])
```

5. Updating Elements

```python
furits = ["apple","orange","gauava"]
furits[1] = "strawberry "
print(furits)
```

6. List Method

append

```python
num = [10,20,30]
num.append(40)
print(num)
```

extend

```python
a = [10,20,30]
b = [40,50,60]
a.extend(b)
print(a)
```

insert

```python
num = [1,3,4]
num.insert(1,10)
print(num)
```

remove

```python
furits = ["apple","banana","guava","kiwi"]
furits.remove("apple")
print(furits)
```

pop()

```python
numbers = [10,20,30,40,50]
print(numbers.pop())
```

clear()

```python
num = [1,2,3,]
num.clear()
print(num)
```

sort()

```python
num = [1,2,10,8,3,]
num.sort()
print(num)
```

descending

```python
num = [1,2,3,4,5,7]
num.sort(reverse = True)
print(num)
```

reverse()

```python
num = [1,2,3]
num.reverse()
print(num)
```

count

```python
num = [1,2,2,2,23,4]
print(num.count(2))
```

index

```python
furits = ["apple","banana","cherry","guava"]
print(furits.index("apple"))
```

7. Nested list

```python
matrix =[
    [1,2],
    [3,4]
]
print(matrix)
```

Loop Through a list

```python
furits = ["banana","apple","gauava"]
for furit in furits:
  print(furit)
```

List Comprehension

```python
square = [x*x for x in range(1,11) ]
print(square)
```

Even numbers

```python
even = [x for x in range(1,21) if x%2 == 0]
print(even)
```
