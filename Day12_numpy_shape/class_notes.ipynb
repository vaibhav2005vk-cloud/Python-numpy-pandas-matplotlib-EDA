Day 12 Numpy Reshape And Flatting

1. reshape()

Changes the shape of an array without changing its data.

```python
import numpy as np
arr = np.arange(1,11)
new_arr = arr.reshape(2,5)
print(new_arr)
```

2. Reshape into 2d array

```python
import numpy as np
arr = np.arange(1,13)
new_arr = arr.reshape(3,4)
print(new_arr)
```

3. Reshape into 3d array

```python
import numpy as np
arr = np.arange(1,25)
new_arr = arr.reshape(2,3,4)
print(new_arr)
```

Using -1 in reshape

```python
import numpy as np
arr = np.arange(1,13)
print(arr.reshape(3,-1))
print(arr.reshape(-1,2))
```

5. flatten()

Convert a multidimensional array into a new 1D array

```python
import numpy as np
arr = np.array([
    [1,2],
    [3,4]
])
flat = arr.flatten()
print(flat)
```

6. ravel()

Also converts an array into 1D, but usually returns a view of the original array.

```python
import numpy as np
arr = np.array([
    [1,2],
    [3,4]
])
ravel = arr.ravel()
print(ravel)
```

7. Difference between ravle() and flatten()

```python
import numpy as np
arr = np.array([[1,2,3],[4,5,6]])
flat = arr.flatten()
ravel= arr.ravel()
flat[0] = 100
ravel[1] = 200
print(flat)
print(ravel)
print(arr)
```

key idea

° flatten() → returns a copy.
° ravel() → usually returns a view, so changes may affect the original array.

8. transpose()

Swap rows and columns

```python
import numpy as np
arr = np.array([[1,2,3],[4,5,6]])
print(arr.T)
```

9. swapaxes()

```python
import numpy as np
arr = np.arange(24).reshape(2,3,4)
new_arr = np.swapaxes(arr,0,1)
print(new_arr)
```

10. resize()

```python
import numpy as np
arr = np.array([[1,2],[3,4]])
arr.resize(2,2)
print(arr)
```
