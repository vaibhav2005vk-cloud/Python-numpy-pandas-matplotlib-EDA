Pratice Question

1. create an array to 1 to 12
2. Reshape it into 3×4.
3. Reshape it into 2×6.
4. Reshape it into 2×2×3
5. Use-1 with reshape

```python
import numpy as np
arr = np.arange(1,13)
new_arr1 = arr.reshape(3,4)
new_arr2 = arr.reshape(2,6)
new_arr3 = arr.reshape(2,2,3)
print(new_arr1)
print(new_arr2)
print(new_arr3)
print(arr.reshape(3,-1))
print(arr.reshape(-1,2))
```
