⭐ Challenge

Write a program that checks whether a given string is a palindrome.
Example:
Input: madam
Output: Palindrome

Input: python
Output: Not Palindrome

```python
text = input("Enter the text") # input
revers_string = text[::-1] # rverse the string
# use condition to chek
if text == revers_string:
  print("palindrome ")
else:
  print("Not a palindrome")
```
