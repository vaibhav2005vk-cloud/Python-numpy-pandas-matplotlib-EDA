📝 Assignment

Create a Python program that:

1. Takes your full name as input

2. Prints:
° Length of the name
° First character
° Last char° acter
° Reversed name
° Name in uppercase
° Name in lowercase
° Title case

3. Counts how many times the letter "a" appears.

4. Replaces all spaces with "-".

5. Checks whether the name starts with "V" and ends with "r".

```python
name = input("Enter the name:") # input name
print(len(name)) # length
print(name[0]) # first
print(name[12]) # last
print(name[::1]) # reverse string
print(name.upper()) # uppercase
print(name.lower()) # lowercase
print(name.title()) # title
print(name.count("a")) # count
print(name.startswith("v"))# startwith
print(name.endswith("t")) # ends with
```
