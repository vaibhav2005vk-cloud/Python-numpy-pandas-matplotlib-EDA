1. What is string

3. Accessing Character (indexing )

```python
txt = "python"
print(txt[0])#p
print(txt[1])#y
print(txt[2])#t
```

Negative  indexing  

```python
text = "python"
print(text[-1])
print(text[-2])
```

4. String slicing

string[start:end:step]

```python
text = "Machine Learning"
print(text[0:7])
print(text[7:])
print(text[:7])
```

5. String length

```python
text = "Artificial intelligence"
print(len(text))
```

6. Common string method

Uppercase

```python
txt = "python"
print(txt.upper())
```

Lowercase

```python
txt = "PYTHON"
print(txt.lower())
```

Title case

```python
txt = "machine learning "
print(txt.title())
```

Captlizie

```python
txt = "python"
print(txt.capitalize())
```

Remove space

```python
txt = " python "
print(txt.strip())
```

Replace

```python
txt = "i like java"
print(txt.replace("java","python"))
```

Join

```python
txt = ["python","is","awesome"]
print("".join(txt))
```

Count

```python
txt = "Machine learning "
print(txt.count('n'))
```
