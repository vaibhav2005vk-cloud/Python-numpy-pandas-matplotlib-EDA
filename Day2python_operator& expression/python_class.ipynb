Day 2 : Python Operator & Expression  

1. Arithmetic Operators

```python
a = 50
b = 30
print(a + b)   # Addition
print(a - b)   # Subtraction
print(a * b)   # Multiplication
print(a / b)   # Division
print(a // b)  # Floor Division
print(a % b)   # Modulus
```

2.Comphrasion Operator

```python
a = 10
b = 20

print(a == b)
print(a != b)
print(a > b)
print(a < b)
print(a >= b)
print(a <= b)
```

3. Assignment Operators

```python
x = 10

x += 5
x -= 2
x *= 3
x /= 2

print(x)
```
