Day 10 Numpy Array Operation

```python
import numpy as np
arr1 = np.array([1,2,3,4,5])
arr2 = np.array([6,7,8,9,10])
```

Addition

```python
import numpy as np
arr1 = np.array([1,2,3,4,5])
arr2 = np.array([6,7,8,9,10])
print(arr1+arr2)
print(np.add(arr1,arr2))
```

3. Subtraction

```python
import numpy as np
arr1 = np.array([1,2,3,4,5])
arr2 = np.array([6,7,8,9,10])
print(arr1-arr2)
print(np.subtract(arr1,arr2))
```

4. Multiplication

```python
import numpy as np
arr1 = np.array([1,2,3,4,5])
arr2 = np.array([6,7,8,9,10])
print(arr1*arr2)
print(np.multiply(arr1,arr2))
```

Division

```python
import numpy as np
arr1 = np.array([6,7,8,9,10])
arr2 = np.array([1,2,3,4,5])
print(arr1/arr2)
print(np.divide(arr1,arr2))
```

Floor division

```python
print(arr1 // arr2)
```

Modules

```python
print(arr1 % arr2)
```

Power

```python
print(arr2 ** 2)
```

Or

```python
print(np.power(arr2, 2))
```

Scaler Operation

```python
print(arr1 + 10)
print(arr1 * 5)
print(arr1 / 2)
```

Square root

```python
print(np.sqrt(arr1))
```

Exponential

```python
print(np.exp(arr2))
```

Logrthim

```python
print(np.log(arr1))
```

Absoulte value

```python
arr = np.array([-10, -20, 30, -40])

print(np.abs(arr))
```

Comparison Operator

```python
print(arr1 > 20)
print(arr1 == 30)
print(arr1 != 40)
```

Aggregation Function  

```python
print(np.sum(arr1))
print(np.mean(arr1))
print(np.max(arr1))
print(np.min(arr1))
print(np.std(arr1))
```
