Pratice Question

1. Create two arrays of five elements each.
2. Add the arrays.
3. Subtract the arrays.
4. Multiply the arrays.
5. Divide the arrays.
6. Find the square of every element.
7. Find the square root of every element.
8. Add 100 to every element.
9. Find the absolute values of a negative array.
10. Find the sum, mean, maximum, minimum, and standard deviation.

```python
import numpy as np
arr1 = np.array([10,20,30,40,50])
arr2 = np.array([1,2,3,4,5])
print(np.add(arr1,arr2)) #Addition
print(np.subtract(arr1,arr2)) # Subtraction
print(np.multiply(arr1,arr2)) # multiplaction
print(np.divide(arr1,arr2)) # divide
print(np.power(arr1,2)) # square
print(np.sqrt(arr1)) # square Root
print(arr1+100)#Add 100 to array1
print(arr2+100)# Add 100 to second array
```

```python
import numpy as np
arr = np.array([-1,-2,-3,-4,-5])
print(np.abs(arr)) #Absoulte value
```

```python
import numpy as np
arr1 = np.array([10,20,30,40,50,60])
print(np.sum(arr1)) # sum
print(np.mean(arr1)) # mean
print(np.max(arr1)) # max
print(np.min(arr1)) # min
print(np.std(arr1)) # std
```
