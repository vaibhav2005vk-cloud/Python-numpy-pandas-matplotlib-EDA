1.Create two 2×2 matrices.

2. Add them
3. Subtract them.
4. Multiply them using np.dot().
5. Find the transpose.
6. Find the determinant.7.7. Multiply them using @.
8. Find the inverse.
9. Find the rank.
10.Find the trace.

```python
import numpy as np
a = ([
    [1,2],
    [3,4]
])
b = np.array([
    [6,7],
    [8,9]
])
print(a+b) # sum
print(a-b) # subtraction
print(np.dot(a,b)) # multiplication
print(a@b) # multiplication
print(np.transpose(a)) # transpose
print(np.linalg.det(a)) # detreamint
print(np.linalg.inv(b))# inverse
print(np.linalg.matrix_rank)
print(np.trace(a)) # trace
```
