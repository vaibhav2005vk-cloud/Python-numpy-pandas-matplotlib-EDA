3. creating arrays

1D Array

```python
import numpy as np
arr = np.array([1,2,3,4,5])
print(arr)
```

2D Array

```python
import numpy as np
arr = np.array([
    [1,2,3],
    [4,5,6]
])
print(arr)
```

3D Array

```python
import numpy as np
arr = np.array([
    [[1,2,3],[3,5,6]],
    [[7,8,9],[10,11,12]]
])
print(arr)
```

4. Array Proprities

```python
import numpy as np
arr = np.array([[1,2,3],[4,5,6]])
print(arr.ndim)
print(arr.shape)
print(arr.size)
print(arr.dtype)
```
