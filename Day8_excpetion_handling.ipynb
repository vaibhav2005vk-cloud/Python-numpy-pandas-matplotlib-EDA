Day 8  Expection Handling in python

What is exception

An exception is an error that happens while the program is running.

```python
a = 10
b = 0
print(a/b)
```

try and except

```python
try:
  a = 10
  b = 0
  print(a/b)
except ZeroDivisionError:
  print("can not divide by zero")
```

Handling multiple exceptions

```python
try:
  num = int(input("enter the no "))
  print(10/num)
except ValueError:
  print("enter a valid intger")
except ZeroDivisionError:
  print("can not divide by zero")
```

Catching any exception

```python
try:
  num = int(input())
  print(10/num)
except Exception as e:
  print("Error",e)
```

else

```python
try:
  n = int(input("enter the no "))
except ValueError:
  print("invalid input")
else:
  print("you enterd num",n)
```
