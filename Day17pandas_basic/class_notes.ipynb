5. Custom index

```python
import pandas as pd
students = pd.Series(
[80, 90, 75],
index=["Rahul", "Priya", "Aman"]
)

print(students)

```

6. Series Attributes

```python
print(students.index)
print(students.value)
print(students.dtype)
print(students.shape)
```

7. Series Operation

```python
import pandas as pd
marks =pd.Series([67,87,89,87,65])
print(marks+10)
print(marks*10)
```

8. Basic Statics

```python
print(marks.sum())
print(marks.mean())
print(marks.max())
print(marks.min())
print(marks.std())
```

9.Bollean filtering

```python
marks = pd.Series([80, 45, 92, 60, 30])

print(marks[marks > 50])
```
