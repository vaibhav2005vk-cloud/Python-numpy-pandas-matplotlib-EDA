Pratice Question

1. Print only the Name column.
2. Print Name and City.
3. Print the first 2 rows.
4. Print the last 2 rows.
5. Filter students with marks above 75.
6. Filter students from Delhi.
7. Filter students from Delhi with marks above 80.
8. Filter students from Delhi or Mumbai.
9. Sort by age.
10. Print the top 2 students by marks.

```python
import pandas as pd
data ={
    "Name" :["Abhay","Vaibhav","Ayush","Aryan","Rishi"],
    "Age": [24,21,13,10,11],
    "Marks" :[98,89,76,74,54],
    "City" :["Dehli","Mumbai","Lucknow","Kanpur","Unnao"]
}
df = pd.DataFrame(data)
print(df["Name"])
print(df[["Name","City"]])
print(df.head(2))
print(df.tail(2))
print(df[df["Marks"]>75])
print(df[df["City"]=="Kanpur"])
print(df[(df["City"]== "Dehli")&(df["Marks"]> 75)])
```
