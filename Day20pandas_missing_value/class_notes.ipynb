Day 20 Pandas Missing value

1. Dataset

```python
import pandas as pd
import numpy as np

data = {
    "Name": ["Aman", "Rahul", "Priya", "Neha", "Vikas", "Riya"],
        "Age": [21, np.nan, 22, 25, np.nan, 26],
            "City": ["Delhi", "Kanpur", np.nan, "Delhi", "Kanpur", "Lucknow"],
                "Salary": [30000, 45000, np.nan, 50000, 40000, np.nan]
                }

                df = pd.DataFrame(data)

                print(df)
```

2. Missing value check

```python
df.isnull()
```

3. Missing value count

```python
df.isnull().sum()
```

4. Missing all sum value

```python
df.isnull().sum().sum()
```

5. Missing wali row

Agar kisi bhi column me missing hai

```python
df[df.isnull().any =(axis = 1)]
```

sirf age missing

```python
df[df["Age"].isnull()]
```

Salary missing

```python
df[df["Salary"].isnull()]
```

Missing value remove krna

```python
df.dropna()
```

original dataframe ko change krna  

```python
df.dropna(inplace = True)
```

7. Specific coloum ke missing row ko remove krna   

```python
df.dropna(subset=["Salary"])
```

8.Missing value fill karana

```python
df["city"] = df["city"].fillna()
```

9. Mean se fill

```python
df["salary"] = df["salary"].fillna(df["salary"].mean())
```

10 . medain se fill

```python
df["Salary"] = df["Salary"].fillna(
      df["Salary"].median()
      )
```
