Pratice question

1. Missing values identify karo.
2. Har column mein missing values count karo.
3. Total missing values find karo.
4. Age ke missing rows find karo.
5. Salary ke missing rows find karo.
6. Department missing values ko "Unknown" se fill karo.
7. Age ko median se fill karo.
8. Salary ko median se fill karo
9. Missing values wali rows remove karke dekho.
10. Final DataFrame mein koi missing value nahi honi chahiye.

```python
import numpy as np
import pandas as pd
df =pd.DataFrame({
    "Name" : ["A","B","C","D","E","F"],
    "Age" : [20,np.nan,24,np.nan,28,30],
    "Salary" : [25000, 30000, np.nan, 45000, np.nan, 60000],
    "Department":["IT", "HR", np.nan, "IT", "HR", np.nan]
})
print(df.isnull())
print(df[df.isnull().any(axis=1)])
print(df.isnull().sum())
print(df[df["Age"].isnull()])
print(df[df["Salary"].isnull()])
print(df["Department"].fillna("Unknown"))
print(df["Age"].fillna(df["Age"].median()))
print(df.dropna())
print(df.dropna(inplace = True))
```
