Pratice Question

1. Department-wise employee count.
2. Department-wise average salary.
3. Department-wise total salary.
4. Department-wise minimum salary.
5. Department-wise maximum salary.
6. City-wise employee count.
7. City-wise average salary.
8. Highest average-salary department.
9. Lowest average-salary department.
10. Department + City-wise average salary.
11. Department-wise mean, min, max.
12. Result ko DataFrame mein convert using reset_index().

Dataset

```python
import pandas as pd
df= pd.DataFrame({
    "Name":
    [ "Aman", "Rahul", "Priya", "Neha",
     "Vikas", "Riya", "Arjun", "Karan",
     "Simran", "Mohit"],
    "Department" : [
          "IT", "HR", "IT", "Finance",
          "HR", "IT", "Finance", "IT",
          "HR", "Finance"
    ],
    "Salary" : [30000, 45000, 35000, 50000,
            40000, 55000, 60000, 48000,
            42000, 52000]
})
print(df.groupby("Department")["Name"].count())
print(df.groupby("Department")["Salary"].mean())
print(df.groupby("Department")["Salary"].sum())
print(df.groupby("Department")["Salary"].max())
print(df.groupby("Department")["Salary"].min())
print(df.groupby("Department")["Name"].count())
print(df.groupby("Department")["Salary"].mean().idxmax())
print(df.groupby("Department")["Salary"].mean().idxmin())
result = (
    df.groupby("Department")["Salary"].mean().
    reset_index()
)
print(result)
```
