Day 21 Pandas groupby() and Aggregation

Aaj tum seekhoge ki large dataset ko groups mein divide karke analysis kaise karte hain.
Example questions:
IT department ki average salary kya hai?
Kis department mein highest salary hai?
Har city mein kitne employees hain?
Department-wise total salary kitni hai?
Iske liye Pandas ka groupby() extremely important hai.

1. Dataset

```python
df = pd.Dataframe({
    "Name" : ["Aman", "Rahul", "Priya", "Neha", "Vikas", "Riya", "Arjun", "Karan"],
    "Department" : ["IT", "HR", "IT", "Finance",
            "HR", "IT", "Finance", "IT"],
    "City" : ["Delhi", "Kanpur", "Lucknow", "Delhi",
            "Kanpur", "Lucknow", "Delhi", "Kanpur"],
    "Salary" : [30000, 45000, 35000, 50000,
            40000, 55000, 60000, 48000
                ]
})
```

2. Basic group by

```python
df.groupby("Department ")
```

Normally humein iske saath aggregation lagani hoti hai.

3. Department wise Average  salary

```python
df.groupby("Department")["Salary"].mean()
```

Department wise total salry

```python
df.groupby('Department')["salary"].sum()
```

ye batyega ki har department employees  ko kitni salry pay krti hai  

5. Department wise employee count

```python
df.groupby("Department")["Name"].count()
```

9. Multiple aggregation

```python
df.groupby("Department")["Salary"].agg([
    "mean", "min", "max", "sum", "count"
])
```

9. Multiple column

```python
df.groupby("Department").agg({
    "Salary" :["mean","max"],
    "Name" : "count"
})
```

10. city wise analysis

```python
df.groupby("City")["Salary"].mean()
```

11. Groupby+ sorting

```python
df.groupby("Department")["Salary"].mean().sort_values(
   ascending=False
)
```

12. GroupBy result lo data frame banaa

```python
result = (df.groupby("Department")["Salary"].mean().reset_index)
print(result)
```

13. Two level groupby

```python
df.groupby(
      ["Department", "City"]
      )["Salary"].mean()
```
