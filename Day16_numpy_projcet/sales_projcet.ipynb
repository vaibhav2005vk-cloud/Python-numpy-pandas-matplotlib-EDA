Capstone Projcet : sales data analysis system

```python
# import numpy
import numpy as np
# Genrate random sale data
np.random.seed(42)
sales = np.random.randint(1000,10000,size = (10,7))
print(sales)
# Explore the data
print("shape",sales.shape)
print("dimensions",sales.ndim)
print("size",sales.size)
print("data type",sales.dtype)
# Total sale of each store
total = np.sum(sales,axis =1)
print(total)
#Average sale
avg = np.mean(sales,axis =1)
print(avg)
# Day wise
print(np.mean(sales,axis = 0))
# Best performance store
best = np.argmax(total)
print(best)
```
