Student Marks Analysis System

🎯 Project Goal
Analyze marks of 10 students in 5 subjects using NumPy.
Topics covered:
Arrays
Indexing & Slicing
Broadcasting
Statistics
Reshaping
Boolean Indexing
Random Module

```python
# import numpy
import numpy as np
# Genreate student mark
# use random_seed to stay same numbers after refreshing
np.random.seed(42)
marks = np.random.randint(35,101,size = (10,5))
print(marks)
#Array information
print("shape",marks.shape)
print("size",marks.shape)
print("dtype",marks.dtype)
print("dimensions",marks.ndim)
# student wise total marks
total = np.sum(marks,axis= 1)
print(total)
# student wise average
average = np.mean(marks,axis = 1)
print(average)
# subjcet wise average
sub_avg = np.mean(marks,axis = 0)
print(sub_avg)
# highest and lowest marks
print("highest",np.max(marks))
print("Lowest",np.min(marks))
# top student
top = np.argmax(total)
print("Top student index",top)
print("Marks",marks[top])
# student above 80
print(marks[marks>80])
# Grace marks
new_mark =marks+5
print(new_mark)
# pass fail analysis
fail = marks<40
print(fail)
# count fail student
print(np.sum(fail))
# sort marks
print(np.sort(marks,axis =1))
```
