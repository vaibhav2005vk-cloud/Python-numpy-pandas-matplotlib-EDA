varaibles length argument(*args)

```python
def total(*number):
  print(sum(number))

total(10,20,30,40)
```

Keyword Variable-Length Arguments (**kwargs)

```python
def details(**info):
      for key, value in info.items():
              print(key, ":", value)

details(name="Vaibhav", age=21, city="Kanpur")
```

Labmda function

```python
square = lambda x : x*x
print(square(5))
```

Recursion  

```python
def factorial(n):
      if n == 1:
        return 1
      return n*factorial(n-1)

print(factorial(5))
```
