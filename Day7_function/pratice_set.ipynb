write a function to find a factorial

```python
def factorial(a):
  fact = 1
  for i in range(1,a+1):
    fact *= i
  print(fact)
factorial(int(input("Enter the no")))
```

```python
def Fibonacci(n):
  a = 0
  b = 1
  for i in range(1,n+1):
    print(a,end ="")
    a,b = b,a+b

Fibonacci(int(input("enter the no")))
```

Write a function using *args to find the sum of any number of values.

```python
def total(*numbers):
  print(sum(numbers))

total(10,20,30,40,50)
```
